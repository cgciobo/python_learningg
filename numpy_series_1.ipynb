```python
# 0D array
a=np.array(45)
print(a)

print()

# 2D array - nested list
D2 = np.array([[2,4,6,8],[1,3,5,7]])
print(D2)
print()

# 0 - se refera la primul array [[0],[1]] si 2 se refera la indexul din array
print(D2[0,2])
print()

# [start:end] or [start:end:step]
print(D2[1,1:4])
```
